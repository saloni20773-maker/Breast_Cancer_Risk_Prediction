```python
#Age distribution
sns.histplot(df['Age at Diagnosis'])
plt.title("Age Distribution")
plt.show()
```

```python
#Survival Count
sns.countplot(x=df['Overall Survival Status'])
plt.title("Survival Status")
plt.show()
```

```python
#Age vs Survival
sns.boxplot(x=df['Overall Survival Status'], y=df['Age at Diagnosis'])
plt.title("Age vs Survival")
plt.show()
```

```python
#Tumor Size vs Survival
sns.boxplot(x=df['Overall Survival Status'], y=df['Tumor Size'])
plt.title("Tumor Size vs Survival")
plt.show()
```

```python
#Numeric-only Pearson correlation heatmap
plt.figure(figsize=(8,6))
sns.heatmap(df.select_dtypes(include='number').corr(), cmap='coolwarm')
plt.title("Correlation Heatmap")
plt.show()
```
